## Công thức dùng trong cả bài

Gọi $C(u, v)$ là tập các phim mà **cả hai** người dùng $u$ và $v$ đã chấm. **Độ tương tự** giữa $u$ và $v$:

$$\text{sim}(u, v) = \frac{1}{\dfrac{1}{|C(u, v)|}\sum_{j \in C(u, v)} |r_{uj} - r_{vj}| \; + \; 0.001}$$

tức là nghịch đảo của *trung bình độ lệch điểm trên các phim chung* (cộng thêm 0.001 để không chia cho 0). Nếu $C(u, v)$ rỗng thì $\text{sim}(u, v) = 0$. Lưu ý $\text{sim}(u, u) = 1/0.001 = 1000$.

**Điểm dự đoán** của người dùng $u$ cho phim $j$ là trung bình có trọng số trên tập $V_j$ các người dùng đã chấm phim $j$ (kể cả chính $u$ nếu $u$ đã chấm):

$$p_{uj} = \frac{\sum_{v \in V_j} \text{sim}(u, v)\, r_{vj}}{\sum_{v \in V_j} \text{sim}(u, v)}$$

Nếu mẫu số bằng 0 (mọi người đã chấm phim $j$ đều có độ tương tự 0 với $u$) thì **không đủ thông tin để dự đoán** và $p_{uj}$ = `NaN`.

Hai lưu ý:
- $p_{uj}$ được tính cho **mọi** phim, kể cả phim $u$ đã chấm (khi đó nó **không** bằng $r_{uj}$). Tên biến `filled_ratings` của các phần sau giữ theo bản cũ, nhưng bạn nhớ là nó ghi đè cả những ô đã có điểm.
- Bài này không đánh giá công thức trên *dự đoán tốt hay dở*. Việc của ta là **tính đúng** nó, và tính **nhanh**, **nhẹ**.

````python
# Đo thời gian B1 cho một batch 32 người dùng (không tính điểm)
try:
    t0 = time.perf_counter()
    with warnings.catch_warnings():
        warnings.simplefilter('ignore', RuntimeWarning)
        _b = ratings[:batch_size]
        _s = 1 / (np.nanmean(np.abs(_b[:, None, :] - ratings[None]),
                             axis=2) + 0.001)
        _s[np.isnan(_s)] = 0
        _w = (~np.isnan(ratings))[None] * _s[:, :, None]
        _w = _w / _w.sum(axis=1, keepdims=True)
        _f = np.nansum(ratings[None] * _w, axis=1)
    t_B1_per_user = (time.perf_counter() - t0) / batch_size
    del _b, _s, _w, _f
    print(f'B1 - broadcasting  : {t_B1_per_user:.3f} giây / 1 người dùng')
    print(f'A  - vòng lặp thuần: {t_A_per_user:.3f} giây / 1 người dùng')
    print(f'B1 nhanh hơn A     : {t_A_per_user / t_B1_per_user:.1f} lần')
except NameError:
    print('Bạn chưa làm xong phần A.')
````

### B3. Viết lại bằng phép nhân ma trận (1.5đ)

Gọi `M` là ma trận 0/1 cùng kích thước với `ratings`, `M[v, j] = 1` nếu người dùng `v` đã chấm phim `j`; gọi `R0` là `ratings` với `NaN` thay bằng 0. Gọi `S` là ma trận độ tương tự **đầy đủ** `n_users` × `n_users`.

**Dự đoán.** Tử số và mẫu số của công thức $p_{uj}$ chính là hai phép nhân ma trận:
$$\text{tử số} = S \cdot R_0, \qquad \text{mẫu số} = S \cdot M.$$

**Độ tương tự.** Số phim chung của mọi cặp là `M @ M.T`. Khó hơn là tổng độ lệch $\sum_{j \in C(u,v)} |r_{uj} - r_{vj}|$. Vì điểm là **số nguyên từ 1 đến 5**, ta có đẳng thức:
$$|a - b| = \sum_{t=1}^{4} \big([a > t] \ \text{xor}\ [b > t]\big)$$
trong đó $[a > t]$ bằng 1 nếu $a > t$ và bằng 0 nếu không. (Ví dụ số và cách kiểm đẳng thức này nằm trong `Example.ipynb`.) Với mỗi ngưỡng `t`, đặt `P_t` là ma trận 0/1, `P_t[v, j] = 1` nếu `v` đã chấm phim `j` **và** điểm lớn hơn `t`. Với hai số 0/1 thì `x xor y = x + y - 2xy`, nên tổng số phim chung mà **đúng một** trong hai người chấm lớn hơn `t` là một biểu thức gồm ba phép nhân ma trận giữa `P_t`, `M` và chuyển vị của chúng. Cộng qua 4 ngưỡng là ra tổng độ lệch.

Viết hai hàm, được dùng vòng lặp `for t in range(1, 5)` và không được dùng vòng lặp nào khác:
