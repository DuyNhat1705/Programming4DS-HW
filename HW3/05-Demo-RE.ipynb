# Regular Expression

```python
import re

def show_regex_match(text, regex):
    '''
    Prints the string with the regex match highlighted.
    
    This function is modified from this source:
    https://www.textbook.ds100.org/ch/12/text_regex.html
    '''
    result = re.sub(f'({regex})', r'\033[1;30;43m\1\033[m', text)
    result = result.replace('\n', '\\n') # Make \n visible
    print(result)
```

So the whole string for <span style="color:red">red</span> text <span style="color:yello">yellow</span> background is:</br>
        `\033[0;31;43mlove\033[0m`
- `\033[0;31;43m`: mark that you start a section with **red** text (31), **yellow** background (43)
- love: the text you want to format
- `\033[0m`: signal that you back to normal color

**b.  Now let back and see the `show_regex_match` function</br>**
This line from the function:

    result = re.sub(f'({regex})', r'\033[1;30;43m\1\033[0m', text)

The matching text get highlight in the same way we see above.  
- `\033[1;30;43m`: `light (1)` `black (30)` text color with `yellow (43)` background
- `\1`: the matching text found by regex. To use this, we must put regex in ground: `({regex})`
- `\033[0m`: back to normal color

#### Task 2: Word Frequency Analysis by Era
Analyzes how frequently certain keywords appear in different time periods
of State of the Union speeches.

Time Periods:
- Early Republic (1790-1850)
- Civil War Era (1850-1900)
- 20th Century (1900-2000)
- 21st Century (2000-2022)

Keywords tracked: war, peace, economy, freedom, liberty, security, terrorism, technology

```python
# Split by speech separator
speeches = text.split('***')

parsed_speeches = []

for speech in speeches:
    speech = speech.strip()
    if not speech:
        continue
    
    # Extract president name (line after "State of the Union Address")
    president_match = re.search(r'State of the Union Address\s*\n\s*(.+?)\s*\n', speech)
    if not president_match:
        continue
    
    president = president_match.group(1).strip()
    
    # Extract date and year
    date_match = re.search(r'([A-Z][a-z]+\s+\d{1,2},\s+(\d{4}))', speech)
    if not date_match:
        continue
    
    date = date_match.group(1)
    year = int(date_match.group(2))
    
    # Extract the speech text (everything after the date)
    speech_start = date_match.end()
    speech_text = speech[speech_start:].strip()
    
    parsed_speeches.append((president, date, year, speech_text))
```

```python
def count_word_occurrences(text, word):
    """
    Count occurrences of a word in text (case-insensitive, whole word only).
   
    Args:
        text: The text to search
        word: The word to count
    
    Returns:
        Number of occurrences
    """
    pattern = r'\b' + re.escape(word) + r'\b'
    matches = re.findall(pattern, text, re.IGNORECASE)
    return len(matches)
```

```python
def count_total_words(text):
    """
    Count total words in text.
    
    Returns:
        Total word count
    """
    words = re.findall(r'\b\w+\b', text)
    return len(words)
```

```python
def categorize_by_era(speeches):
    """
    Categorize speeches by historical era based on year.
    
    Returns:
        Dictionary with era names as keys and lists of speeches as values
    """
    eras = {
        'Early Republic (1790-1850)': [],
        'Civil War Era (1850-1900)': [],
        '20th Century (1900-2000)': [],
        '21st Century (2000-2022)': []
    }
    
    for president, date, year, text in speeches:
        if 1790 <= year < 1850:
            eras['Early Republic (1790-1850)'].append((president, date, year, text))
        elif 1850 <= year < 1900:
            eras['Civil War Era (1850-1900)'].append((president, date, year, text))
        elif 1900 <= year < 2000:
            eras['20th Century (1900-2000)'].append((president, date, year, text))
        elif 2000 <= year <= 2022:
            eras['21st Century (2000-2022)'].append((president, date, year, text))
    
    return eras
```

```python
def analyze_word_frequency_by_era(speeches, keywords):
    """
    Analyze word frequency for given keywords across different eras.
    
    Args:
        speeches: List of parsed speeches
        keywords: List of keywords to track
    
    Returns:
        Dictionary with analysis results
    """
    # Categorize speeches by era
    eras = categorize_by_era(speeches)
    
    results = {}
    
    for era_name, era_speeches in eras.items():
        # Combine all text from this era
        combined_text = ' '.join([text for _, _, _, text in era_speeches])
        total_words = count_total_words(combined_text)
        
        # Count each keyword
        keyword_counts = {}
        keyword_frequencies = {}
        
        for keyword in keywords:
            count = count_word_occurrences(combined_text, keyword)
            keyword_counts[keyword] = count
            
            # Calculate frequency per 1000 words
            frequency_per_1000 = (count / total_words * 1000) if total_words > 0 else 0
            keyword_frequencies[keyword] = frequency_per_1000
        
        results[era_name] = {
            'speech_count': len(era_speeches),
            'total_words': total_words,
            'keyword_counts': keyword_counts,
            'keyword_frequencies': keyword_frequencies
        }
    
    return results
```

```python
def print_results(results, keywords):
    """
    Print analysis results in a formatted way.
    """
    print("=" * 80)
    print("WORD FREQUENCY ANALYSIS BY ERA")
    print("=" * 80)
    print()
    
    # Print summary for each era
    for era_name, data in results.items():
        print(f"\n{era_name}")
        print("-" * 80)
        print(f"Number of speeches: {data['speech_count']}")
        print(f"Total words: {data['total_words']:,}")
        print()
        print(f"{'Keyword':<15} {'Count':>10} {'Per 1000 words':>20}")
        print("-" * 50)
        
        for keyword in keywords:
            count = data['keyword_counts'][keyword]
            freq = data['keyword_frequencies'][keyword]
            print(f"{keyword:<15} {count:>10} {freq:>20.2f}")
        print()
    
    # Print comparison table
    print("\n" + "=" * 80)
    print("COMPARISON TABLE: Frequency per 1000 words")
    print("=" * 80)
    print()
    
    # Header
    era_names = list(results.keys())
    header = f"{'Keyword':<15}"
    for era in era_names:
        # Shorten era names for table
        short_era = era.split('(')[0].strip()[:12]
        header += f"{short_era:>15}"
    print(header)
    print("-" * (15 + 15 * len(era_names)))
    
    # Data rows
    for keyword in keywords:
        row = f"{keyword:<15}"
        for era in era_names:
            freq = results[era]['keyword_frequencies'][keyword]
            row += f"{freq:>15.2f}"
        print(row)
    
    print()
```

```python
def find_top_keywords_per_era(results, keywords):
    """
    Find which keywords were most prominent in each era.
    """
    print("\n" + "=" * 80)
    print("TOP KEYWORDS PER ERA (by frequency)")
    print("=" * 80)
    print()
    
    for era_name, data in results.items():
        # Sort keywords by frequency
        sorted_keywords = sorted(
            keywords,
            key=lambda k: data['keyword_frequencies'][k],
            reverse=True
        )
        
        print(f"{era_name}:")
        for i, keyword in enumerate(sorted_keywords[:3], 1):
            freq = data['keyword_frequencies'][keyword]
            count = data['keyword_counts'][keyword]
            print(f"  {i}. {keyword}: {freq:.2f} per 1000 words (total: {count})")
        print()
```

```python
def analyze_trends(results, keywords):
    """
    Analyze trends over time for each keyword.
    """
    print("\n" + "=" * 80)
    print("TREND ANALYSIS")
    print("=" * 80)
    print()
    
    era_order = [
        'Early Republic (1790-1850)',
        'Civil War Era (1850-1900)',
        '20th Century (1900-2000)',
        '21st Century (2000-2022)'
    ]
    
    for keyword in keywords:
        print(f"\n{keyword.upper()}:")
        frequencies = [results[era]['keyword_frequencies'][keyword] for era in era_order]
        
        # Calculate trend
        if frequencies[0] > 0:
            change = ((frequencies[-1] - frequencies[0]) / frequencies[0]) * 100
        else:
            change = float('inf') if frequencies[-1] > 0 else 0
        
        # Find peak era
        max_freq = max(frequencies)
        peak_era_idx = frequencies.index(max_freq)
        peak_era = era_order[peak_era_idx].split('(')[0].strip()
        
        print(f"  Peak usage: {peak_era} ({max_freq:.2f} per 1000 words)")
        
        if change != float('inf'):
            trend = "increased" if change > 0 else "decreased"
            print(f"  Overall trend: {trend} by {abs(change):.1f}%")
        else:
            print(f"  Overall trend: New term (not used in Early Republic)")
        
        # Show progression
        print(f"  Progression: ", end="")
        for i, (era, freq) in enumerate(zip(era_order, frequencies)):
            short_era = era.split('(')[1].split('-')[0] if '(' in era else era[:4]
            print(f"{short_era}: {freq:.2f}", end="  ")
        print()
```
