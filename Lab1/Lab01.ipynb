````python
# Ban ngay tho - doc het ca file vao bo nho. Ban khong sua cell nay.
def top_committers_naive(path, year):
    with open(path, 'r', encoding='utf-8-sig') as f:
        text = f.read()
    lines = text.split('\n')
    header = parse_tsv_line(lines[0])
    i_name = header.index('AuthorName')
    i_date = header.index('AuthorDate')
    count = {}
    for line in lines[1:]:
        if not line:
            continue
        row = parse_tsv_line(line)
        if int(row[i_date].split()[3]) != year:
            continue
        count[row[i_name]] = count.get(row[i_name], 0) + 1
    return dict(sorted(count.items(), key=lambda kv: (-kv[1], kv[0])))
````
