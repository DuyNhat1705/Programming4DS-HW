````python
print(8 % 5, -8 % 5, 8 % -5, -8 % -5)      # => 3 2 -2 -3
print(8 // 5, -8 // 5, 8 // -5, -8 // -5)  # => 1 -2 -2 1
````

**TODO 1.1**: Giải thích vì sao phép `%` của Python cho ra các kết quả như trên, chẳng hạn vì sao `-8 % 5` bằng `2` (trong khi C/C++ cho ra `-3`).

Gợi ý: Python luôn bảo đảm `(a // b) * b + (a % b) == a`.

YOUR ANSWER HERE

Từ đẳng thức `(a // b) * b + (a % b) == a`, ta có được công thức của phép chia lấy dư `%` trong Python `a % b == a - (a // b) * b`, mà phép chia lấy nguyên `//` trong Python luôn đảm bảo kết quả `a // b` được **làm tròn xuống (floor)** nên xuất hiện các kết quả như trên khi thực hiện phép tính. Trong khi ở C++, công thức của toán tử `%` được định nghĩa là `a % b == a - (a / b) * b`, ở đó phép chia thông thường `/` sẽ **cắt phần thập phân, lấy phần nguyên của kết quả** `a / b` chứ không làm tròn xuống (floor) như Python.

Phân tích các phép chia lấy dư trong Python:
- 8 % 5 = 8 - ( 8 // 5 ) * 5 = 8 - 1 * 5 = 3
- -8 % 5 = -8 - (-8 // 5) * 5 = -8 - (-2) * 5 = 2
- 8 % -5 = 8 - ( 8 // (-5)) * (-5) = 8 - (-2) * (-5) = -2
- -8 % -5 = -8 - (-8 // (-5)) * (-5) = -8 - 1 * (-5) = -3

Phân tích phép chia lấy dư trong C++:
- -8 % 5 =  -8 - (-8 / 5) * 5 = -8 - (-1) * 5 = -3
- 8 % -5 = 8 - (8 / (-5)) * (-5) = 8 - (-1) * (-5) = 3
- -8 % -5 = -8 - (-8 / (-5)) * (-5) = -8 - 1 * (-5) = -3

**Task 3.10**: Viết hàm `fizzbuzz(n)` **trả về** (không phải in ra) một chuỗi:
- `"Fizz"` nếu `n` chia hết cho 3
- `"Buzz"` nếu `n` chia hết cho 5
- `"FizzBuzz"` nếu `n` chia hết cho cả 3 và 5
- Ngược lại thì trả về chính số `n` dưới dạng chuỗi (vd, `fizzbuzz(7)` trả về `"7"`)

Đoạn code bên dưới hàm dùng hàm của bạn để in ra kết quả cho các số từ 1 đến 30.

````python
def fizzbuzz(n):
    """Trả về chuỗi FizzBuzz ứng với số nguyên dương n."""
    # YOUR CODE HERE
    
    # Task 3.10:
    if n % 3 == 0 and n % 5 == 0:
        return "FizzBuzz"
    elif n % 3 == 0:
        return "Fizz"
    elif n % 5 == 0:
        return "Buzz"
    else:
        return f"{n}"
 

# In thử kết quả cho các số từ 1 đến 30
for i in range(1, 31):
    print(fizzbuzz(i), end=' ')
````

````python
# TEST
assert fizzbuzz(1) == '1'
assert fizzbuzz(3) == 'Fizz'
assert fizzbuzz(5) == 'Buzz'
assert fizzbuzz(15) == 'FizzBuzz'
assert fizzbuzz(98) == '98'
assert [fizzbuzz(i) for i in range(1, 16)].count('Fizz') == 4
seq = ' '.join(fizzbuzz(i) for i in range(1, 101))
assert adler32(seq.encode()) == 3362683043
````

**Task 4.14**: Viết hàm `apply_to_all(func, lst)` nhận vào một hàm `func` và một list `lst`, trả về một **list mới** gồm kết quả của việc áp dụng `func` lên từng phần tử của `lst`. Hàm <font color=red>không được sửa</font> `lst`.

````python
def apply_to_all(func, lst):
    """Áp dụng func lên từng phần tử của lst, trả về list mới."""
    # YOUR CODE HERE
    
    # Task 4.14:
    return [func(x) for x in lst]


# Thử với hàm polynomial của bạn
test_list = [0, 1, 2, 3, 4, 5]
result = apply_to_all(polynomial, test_list)
print(f"Applied to {test_list}: {result}")
````

````python
# TEST
expected = [polynomial(x) for x in test_list]
assert apply_to_all(polynomial, test_list) == expected
assert isinstance(apply_to_all(abs, [-1, 2]), list)  # phải là list
assert apply_to_all(str.upper, ['a', 'b']) == ['A', 'B']
assert apply_to_all(len, []) == []
lst = [3, -4]
new_lst = apply_to_all(abs, lst)
assert new_lst == [3, 4]
assert lst == [3, -4] and new_lst is not lst  # không sửa list gốc
````

**Task 4.15**: Viết hàm `calculate_grade(score)` nhận vào điểm số nguyên `score` (từ 0 đến 100) và trả về điểm chữ:
- A: 90 – 100
- B: 80 – 89
- C: 70 – 79
- D: 60 – 69
- F: 0 – 59

````python
def calculate_grade(score):
    """Trả về điểm chữ ('A'..'F') ứng với điểm số score (0..100)."""
    # YOUR CODE HERE
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    return 'F'


# Thử
print(calculate_grade(95))  # => 'A'
print(calculate_grade(82))  # => 'B'
print(calculate_grade(55))  # => 'F'
````

````python
# TEST
expected = {100: 'A', 90: 'A', 89: 'B', 80: 'B', 79: 'C',
            70: 'C', 69: 'D', 60: 'D', 59: 'F', 0: 'F'}
for score, grade in expected.items():
    assert calculate_grade(score) == grade, f'calculate_grade({score}) sai'
grades = ''.join(calculate_grade(s) for s in range(101))
assert adler32(grades.encode()) == 2002197262
````

**Task 5.16**: Viết hàm `save_student_data(filename, student_dict)` ghi dictionary `student_dict` ra file `filename` theo định dạng **JSON**.

````python
def save_student_data(filename, student_dict):
    """Ghi dictionary student_dict ra file JSON tên filename."""
    import json
    # YOUR CODE HERE
    
    # Task 5.16:
    with open(filename, "w", encoding="utf-8") as file:
        file.write(json.dumps(student_dict, ensure_ascii=False))
````

````python
# TEST
sv_test = {"name": "Nguyễn Văn A", "id": "24120001", "gpa": 3.5,
           "courses": ["P4DS", "NMKHDL"], "graduated": False}
save_student_data("test_student.json", sv_test)
with open("test_student.json", encoding="utf-8") as f:
    assert json.load(f) == sv_test  # file phải là JSON hợp lệ
````

**Task 5.17**: Viết hàm `load_student_data(filename)` đọc file JSON do hàm ở trên ghi ra và trả về dictionary.

````python
def load_student_data(filename):
    """Đọc file JSON tên filename, trả về dictionary."""
    import json
    # YOUR CODE HERE
    
    # Task 5.17:
    with open(filename, "r", encoding="utf-8") as file:
        res = json.load(file)
    return res


# Thử hai hàm của bạn
test_student = {
    "name": "John Doe",
    "id": "12345",
    "major": "Computer Science",
    "gpa": 3.5
}
save_student_data("test_student.json", test_student)
loaded_student = load_student_data("test_student.json")
print("Loaded student:", loaded_student)
````

````python
# TEST
sv_test = {"name": "Trần Thị B", "id": "24120002", "gpa": 3.25,
           "courses": ["P4DS"], "scores": {"Lab00": 10.0}}
save_student_data("test_student.json", sv_test)
loaded = load_student_data("test_student.json")
assert isinstance(loaded, dict)
assert loaded == sv_test
assert loaded is not sv_test
````
